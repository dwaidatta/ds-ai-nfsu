### Operator Precedence and Associativity

**Precedence** decides which operator is evaluated first when multiple operators appear in one expression. **Associativity** decides the evaluation direction (left-to-right or right-to-left) when operators have the same precedence.

| Precedence (High → Low) | Operator(s) | Associativity | Notes |
|---|---|---|---|
| 1 | `()` | Left to right | Parentheses evaluated first, always |
| 2 | `**` | Right to left | Exponent is evaluated right-to-left |
| 3 | `+x`, `-x` (unary) | Right to left | Unary plus/minus, e.g. `-5` |
| 4 | `*`, `/`, `//`, `%` | Left to right | Multiplication group |
| 5 | `+`, `-` (binary) | Left to right | Addition/subtraction group |
| 6 | `==`, `!=`, `>`, `>=`, `<`, `<=` | Left to right | Comparison operators |
| 7 | `is`, `is not`, `in`, `not in` | Left to right | Identity and membership checks |
| 8 | `not` | Right to left | Logical NOT |
| 9 | `and` | Left to right | Logical AND |
| 10 | `or` | Left to right | Logical OR |

**Quick Example:** In the expression `2 + 3 * 2 ** 2`, Python evaluates `**` first (`2 ** 2 = 4`), then `*` (`3 * 4 = 12`), then `+` (`2 + 12 = 14`). The final result is `14`.

Lastly, let's understand identity operators using `is` and `is not`. These check whether two variables point to the exact same object in memory, not just whether their values are equal.

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print(list1 == list2)
print(list1 is list2)
print(list1 is list3)
````

`list1` and `list2` have the same values, so `list1 == list2` is `True`. But they are two separate objects created independently in memory, so `list1 is list2` is `False`.

On the other hand, `list3` was assigned directly from `list1`, meaning it points to the same object in memory, so `list1 is list3` gives `True`.

### Loops

Till now, whatever code we wrote executed line by line, once each. But in real life, we often need to repeat an action multiple times.

Suppose you want to print numbers from 1 to 100. Would you write 100 print statements? Certainly not. This is where **loops** come in.

A loop allows us to repeat a block of code multiple times without writing it again and again. Python provides two main types of loops:

- The `for` loop, used when we want to repeat an action a specific number of times or over a sequence of values.
- The `while` loop, used when we want to repeat an action as long as a certain condition remains true.

#### The `for` loop

Before jumping into code, let's understand a few important terms.

A `for` loop is generally used to go through (or **iterate** over) a sequence of values, one at a time. This sequence could be a list, a string, a range of numbers, or any other collection of values.

The variable used inside a `for` loop to hold each value during the iteration is called the **iterator variable** (also sometimes called the loop variable). It automatically takes each value from the sequence, one after another, until the sequence is exhausted.

The basic syntax looks like this:

```python
for iterator_variable in sequence:
    # block of code to repeat
```

Notice the colon `:` at the end of the `for` line, and the indentation of the code below it. This indentation tells Python which lines belong inside the loop. This is similar to how we saw indentation matter in conditional statements.

Often, we don't already have a ready sequence to loop over, but just want to repeat something a fixed number of times, like counting from 1 to 5. For this, Python gives us a built-in function called `range()`.

The `range()` function generates a sequence of numbers, which we can then iterate over using a `for` loop.

It can be used in three ways:

- `range(stop)`: generates numbers starting from `0` up to (but not including) `stop`.
- `range(start, stop)`: generates numbers starting from `start` up to (but not including) `stop`.
- `range(start, stop, step)`: generates numbers starting from `start` up to (but not including) `stop`, increasing (or decreasing) by `step` each time.

Let's understand all of this with examples now.

````python
for i in range(5):
  print(i)
````

Here, `range(5)` generates numbers from `0` to `4` (5 numbers in total, but not including 5). The variable `i` is our iterator variable — it takes the value `0` first, runs the loop body (`print(i)`), then takes `1`, runs again, and so on until it has taken the value `4`.

Notice that the sequence starts from `0` by default, not `1`.

````python
for i in range(2, 6):
  print(i)
````

This time we used `range(2, 6)`, which means the sequence starts from `2` and goes up to (but not including) `6`. So the values taken by `i` are `2, 3, 4, 5`.

````python
for i in range(1, 10, 2):
  print(i)
````

Here we added a third value, `2`, which is the **step**. Instead of increasing by `1` each time, `i` now increases by `2`. So starting from `1`, we get `1, 3, 5, 7, 9` — all odd numbers below `10`.

The step value can also be negative, which allows us to count backward.

````python
for i in range(10, 0, -2):
  print(i)
````

Here, `i` starts at `10` and decreases by `2` each time, stopping just before it reaches `0`. So the values printed are `10, 8, 6, 4, 2`.

The `for` loop is not limited to just numbers from `range()`. It can iterate over other sequences too, like a list or even a string, since a string is just a sequence of characters.

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
  print(fruit)
````

Here, our iterator variable is named `fruit` (we can name it anything meaningful). It takes each value from the `fruits` list one by one — first `"apple"`, then `"banana"`, then `"cherry"` — and the loop body prints it each time.

````python
name = "Alice"

for letter in name:
  print(letter)
````

Since a string is a sequence of characters, the `for` loop goes through it one character at a time. So `letter` takes the value `A`, then `l`, then `i`, then `c`, then `e`, printing each on a new line.

#### The `while` loop

Unlike the `for` loop, which is generally used when we know how many times we want to repeat something (or already have a sequence to go through), the `while` loop is used when we want to repeat an action **as long as a condition remains true**.

The basic syntax looks like this:

```python
while condition:
  # block of code to repeat
```

The condition is checked before every repetition. If the condition is `True`, the loop body runs. Once the loop body finishes, the condition is checked again. This continues until the condition becomes `False`, at which point the loop stops.

It is important to make sure that something inside the loop eventually makes the condition `False`. Otherwise, the loop will run forever, known as an **infinite loop**.

Let's see this with examples.

````python
count = 1

while count <= 5:
  print(count)
  count += 1
````

Here, `count` starts at `1`. The condition `count <= 5` is checked before each run of the loop body.

As long as this condition is `True`, the loop prints the current value of `count` and then increases it by `1` using `count += 1`. Once `count` becomes `6`, the condition `count <= 5` becomes `False`, and the loop stops.

Notice that `count += 1` is essential here. Without it, `count` would always remain `1`, the condition would always stay `True`, and we would end up with an infinite loop.

````python
num = 20

while num > 0:
  print(num)
  num -= 5
````

This time, `num` starts at `20` and the loop runs as long as `num` is greater than `0`. Each time, we print `num` and then reduce it by `5` using `num -= 5`. So the values printed are `20, 15, 10, 5`. Once `num` becomes `0`, the condition `num > 0` becomes `False`, and the loop ends.

#### The `break` and `continue` statements

So far, our loops ran from start to finish: a `for` loop went through every item, and a `while` loop kept going until its condition became `False`. But sometimes we want more control. We may want to stop a loop early because we already found what we were looking for, or skip one particular item and carry on with the rest.

Python gives us two statements for this:

- `break` – stops the loop immediately and moves on to the code after the loop.
- `continue` – skips the rest of the current repetition and jumps straight to the next one.

Think of a teacher checking a pile of answer sheets. If she finds a sheet with copied answers and decides to stop checking altogether, that is a `break`. If she simply sets that one sheet aside and picks up the next, that is a `continue`.

Let's see them one by one.

Let's start with `break`. Here we want to print numbers from `1` to `9`, but stop as soon as we reach `5`.

````python
for i in range(1, 10):
  if i == 5:
    break
  print(i)

print("Loop ended")
````

Here, the loop goes through `1, 2, 3, ...` one by one. For each value, the `if` statement checks whether `i` is equal to `5`.

- For `1` to `4`, the condition is `False`, so the value is printed.
- When `i` becomes `5`, the condition is `True` and `break` executes. The loop stops right there, so `5` is never printed, and neither are `6` to `9`.

Notice that the last `print` is not indented, so it is outside the loop. It runs after the loop is stopped by `break`.

A very common use of `break` is to search for something in a collection and stop once it is found. There is no point checking the remaining items after that.

````python
marks = [45, 78, 92, 34, 60]

for m in marks:
  if m < 40:
    print("First failing mark:", m)
    break
````

The loop checks each mark in the list. The first mark below `40` is `34`, so we print it and `break`. The last mark, `60`, is never even looked at, because we already have our answer.

*If we had not used `break`, the loop would keep going through the remaining marks for no reason. For a list of 5 values that hardly matters, but for a list of a million values it saves a lot of work.*

The `break` statement works the same way in a `while` loop. It is especially handy with `while True:`, a loop whose condition is always true and which would run forever on its own. We use `break` to provide the exit.

````python
count = 1

while True:
  print(count)
  if count == 3:
    break
  count += 1
````

Here, the condition `True` never becomes `False`, so the loop can only end through `break`. We print `count`, and once it reaches `3` the `if` condition is satisfied and the loop stops. Otherwise, `count` is increased by `1` and the loop repeats.

Now let's move to `continue`. Here we print numbers from `1` to `5` but skip the number `3`.

````python
for i in range(1, 6):
  if i == 3:
    continue
  print(i)
````

When `i` is `3`, the `continue` statement executes. The rest of the loop body (the `print`) is skipped for this round, and the loop jumps to the next value, `4`. That is why `3` is missing from the output.

The difference from `break` is important: `continue` does **not** end the loop. It only skips the current repetition, and the loop carries on with the remaining items.

`continue` also works in a `while` loop, but here we need to be careful. Let's see an example.

````python
count = 0

while count < 5:
  count += 1
  if count == 3:
    continue
  print(count)
````

This gives the same output as the `for` loop above. Notice where `count += 1` is placed: **before** the `if` statement.

When `continue` executes in a `while` loop, it jumps straight back to checking the condition. If `count += 1` were placed after the `continue`, it would be skipped when `count` is `3`. Then `count` would stay `3` forever, the condition would always be `True`, and we would end up with an infinite loop. So always make sure the value that controls the loop is updated before `continue` is used.

To summarise:

| Statement | What it does | Loop continues? |
|---|---|---|
| `break` | Exits the loop completely | No, the loop ends |
| `continue` | Skips the rest of the current repetition | Yes, with the next repetition |

Both are usually used together with an `if` statement, so that they only trigger when a certain condition is met.

## Working with Collections of Data

Till now, we have mostly worked with single values stored in variables, like a number or a name. But real-world data rarely comes as a single value. We often deal with a collection of values together, like a list of student names, or a set of marks.

Python gives us several built-in data structures to store collections of values. We will look at four of them:

- **Strings** – a sequence of characters
- **Lists** – an ordered, changeable collection of values
- **Tuples** – an ordered, unchangeable collection of values
- **Dictionaries** – a collection of key-value pairs

All four of these are called **sequences** or **collections**, since they hold more than one value together. Before we explore each one individually, let's understand one function that works across all of them: the `len()` function.

### The `len()` function

Whenever we work with a collection of values, one of the most common things we want to know is: how many values does it hold?

The `len()` function (short for **length**) is a built-in Python function that returns the number of items present in a collection. It works on strings, lists, tuples, and dictionaries alike, though what it "counts" depends slightly on the structure, as we will see.

The syntax is simple:

```python
len(collection)
```

Let's see this in action across different types of collections, before we study each of them in detail.

````python
name = "Alice"
marks = [45, 78, 92, 34, 60]
coordinates = (10, 20, 30)
student = {"name": "Alice", "age": 20}

print(len(name))
print(len(marks))
print(len(coordinates))
print(len(student))
````

- `len(name)` gives `5`, since the string `"Alice"` has 5 characters.
- `len(marks)` gives `5`, since the list has 5 elements.
- `len(coordinates)` gives `3`, since the tuple has 3 elements.
- `len(student)` gives `2`, since the dictionary has 2 key-value pairs. Note that `len()` counts the number of *pairs*, not individual keys and values separately.

Keep the `len()` function in mind as we go through each structure below. We will use it often, especially while looping, to know exactly how many times we need to repeat an action.

### Strings

We have already used strings before, like `"Alice"` or `"Hello"`. A string is simply a sequence of characters enclosed in quotes, and it can represent anything from a single word to entire paragraphs of text.

Since a string is a sequence, each character inside it occupies a specific position, called an **index**. Indexing in Python starts from `0`, not `1`. So in the string `"Alice"`, the character `A` is at index `0`, `l` is at index `1`, and so on, up to `e` at index `4` (since the string has 5 characters in total, matching what `len()` told us earlier).

We can also use **negative indexing** to access characters counting from the end of the string. The last character is at index `-1`, the second-last at `-2`, and so on. This is especially handy when we do not know the exact length of a string but still want to access its last few characters.

Let's see this with an example.

````python
name = "Alice"

print(name[0])
print(name[4])
print(name[-1])
````

Here, `name[0]` gives us `A`, the first character. `name[4]` gives us `e`, the last character counted from the front. `name[-1]` also gives us `e`, but this time counted from the back. Since `len(name)` was `5`, the valid forward indices range from `0` to `4`.

We can also access a portion of a string using **slicing**. Slicing uses the format `string[start:stop]`, which gives us characters from index `start` up to (but not including) `stop`.

````python
word = "Programming"

print(word[0:4])
print(word[4:])
print(word[:4])
````

`word[0:4]` gives us characters from index `0` to `3`, which is `"Prog"`. `word[4:]` means starting from index `4` all the way to the end, giving `"ramming"`. `word[:4]` means starting from the beginning up to index `3`, giving `"Prog"` again, since leaving out the start defaults to `0`.

Since a string is a sequence, we can iterate over it character by character using a `for` loop. We can even combine this with `len()` and `range()` to loop using index positions instead of directly over characters, which is useful when we need to know the position of each character too.

````python
word = "Data"

for letter in word:
  print(letter)

for i in range(len(word)):
  print(i, word[i])
````

The first loop directly iterates over each character of `"Data"`. The second loop instead iterates over index numbers generated by `range(len(word))`, which gives `0, 1, 2, 3` since `len(word)` is `4`. Using `word[i]` inside the loop, we fetch the character at each index, along with the index itself. This pattern of using `range(len(...))` is very common when both the position and the value are needed.

One important thing to remember about strings is that they are **immutable**, meaning once a string is created, its individual characters cannot be changed directly. Let's actually try to do this and see what happens.

````python
name = "Alice"
name[0] = "M"
````

This gives us a `TypeError: 'str' object does not support item assignment`. This confirms that strings do not allow us to modify a character at a given index directly.

So how do string methods like `upper()` or `replace()` appear to "change" the string then? They don't actually modify the original string — they create and return a brand new string instead. Let's verify this.

````python
name = "Alice"
new_name = name.upper()

print(name)
print(new_name)
````

Notice that `name` still holds `"Alice"`, completely unchanged. The `upper()` method created a new string `"ALICE"` and we stored it separately in `new_name`. This is the behavior of every string method — the original string is never touched.

### Lists

A list is an ordered collection of values, and unlike a string, it can hold values of different datatypes together in a single collection. A list is defined using square brackets `[ ]`, with values separated by commas.

````python
marks = [45, 78, 92, 34, 60]
print(marks)
print(len(marks))

mixed = ["Alice", 25, True, 5.5]
print(mixed)
````

Here, `marks` is a list of integers, and `len(marks)` confirms it has `5` elements. The `mixed` list shows that a single list can hold a string, an integer, a boolean, and a float all together, something a string cannot do since it only holds characters.

Just like strings, lists also support indexing, negative indexing, and slicing, since a list is also a sequence.

````python
marks = [45, 78, 92, 34, 60]

print(marks[0])
print(marks[-1])
print(marks[1:4])
````

`marks[0]` gives the first value `45`. `marks[-1]` gives the last value `60`. `marks[1:4]` gives a new list containing values from index `1` to `3`, which is `[78, 92, 34]`.

Unlike strings, lists are **mutable**, meaning we can change their content after creation, without needing to create a new list. Let's actually try modifying a list directly and confirm this works, unlike what happened with strings.

````python
marks = [45, 78, 92, 34, 60]
marks[0] = 50

print(marks)
````

Here, we directly changed the value at index `0` from `45` to `50`, and the original `marks` list itself gets updated in place, unlike what we saw with string methods.

We can iterate over a list using a `for` loop, and just like with strings, we can also combine it with `len()` and `range()` when we need index positions.

````python
marks = [45, 78, 92, 34, 60]

for mark in marks:
  print(mark)

for i in range(len(marks)):
  print(i, marks[i])
````

The first loop goes through each value directly. The second loop goes through index positions from `0` to `len(marks) - 1`, letting us access both the position and the value at each step using `marks[i]`.

### Tuples

A tuple is an ordered collection of values, very similar to a list. Just like a list, it can hold values of different datatypes together, and each value has a fixed position (index).

The key difference is that a tuple is **immutable**, meaning once created, its values cannot be changed, added, or removed. A tuple is defined using round brackets `( )`, with values separated by commas.

Tuples are generally used when we want to store a fixed collection of values that should not change throughout the program, such as coordinates, days of the week, or fixed configuration values.

````python
coordinates = (10, 20)
print(coordinates)
````

````python
info = ("Alice", 25, "Engineer")
print(info)
print(len(info))
````

Here, `coordinates` is a tuple holding two integer values, and `info` is a tuple holding a mix of a string, an integer, and another string. `len(info)` gives us `3`, confirming the tuple holds three values, just as `len()` worked for lists and strings.

A tuple with a single value needs a comma after it, otherwise Python will not treat it as a tuple. Let's see this.

````python
single_value = (5)
print(type(single_value))

single_tuple = (5,)
print(type(single_tuple))
````

Notice that `single_value` is actually just an `int`, since without a comma, Python treats `(5)` as a regular value inside parentheses, not a tuple. Adding a comma after `5` as in `(5,)` is what actually makes it a tuple.

Just like lists and strings, tuples support **indexing**, **negative indexing**, and **slicing**, since a tuple is also a sequence.

````python
info = ("Alice", 25, "Engineer")
print(info[0])
print(info[-1])
print(info[0:2])
````

`info[0]` gives `"Alice"`, the first value. `info[-1]` gives `"Engineer"`, the last value counted from the back. `info[0:2]` gives a new tuple `("Alice", 25)`, containing values from index `0` to `1`.

We can also iterate over a tuple using a `for` loop, since it is a sequence, just like we did with lists and strings.

````python
info = ("Alice", 25, "Engineer")

for item in info:
  print(item)
````

Now let's turn our attention to the immutability of tuples. Let's actually try to modify a tuple and see what happens, just like we did with strings.

````python
info = ("Alice", 25, "Engineer")
info[1] = 26
````

This gives us a `TypeError: 'tuple' object does not support item assignment`. This confirms the core difference between a list and a tuple — a list allows this kind of direct modification, but a tuple does not.

This is exactly why tuples are preferred when we want to store a fixed collection of values that should never accidentally change during the program.

Since tuples cannot be modified, they only support two built-in methods.

| Method | Syntax | Arguments | Description | Example |
|---|---|---|---|---|
| `count()` | `tuple.count(value)` | `value` (mandatory) | Counts number of occurrences of `value` in the tuple | `(1,2,2,3).count(2)` → `2` |
| `index()` | `tuple.index(value)` | `value` (mandatory) | Returns the index of the first occurrence of `value` | `(1,2,3).index(3)` → `2` |

Let's try out both of these directly.

````python
numbers = (10, 20, 10, 30, 10)
print(numbers.count(10))
````

````python
numbers = (10, 20, 10, 30, 10)
print(numbers.index(30))
````

`count(10)` tells us how many times the value `10` appears in the tuple, which is `3`. `index(30)` tells us the position at which the value `30` first appears, which is index `3`.

### Dictionaries

A dictionary is different from strings, lists, and tuples. Instead of storing values at an ordered position (index), it stores data as **key-value pairs**. Each value is associated with a unique key, and we access the value using its key rather than a position.

A dictionary is defined using curly brackets `{ }`. Inside, each key and its corresponding value are written together separated by a colon `:`, and different key-value pairs are separated by commas.

The basic syntax looks like this:

```python
dictionary_name = {key1: value1, key2: value2}
```

Dictionaries are especially useful when we want to label our data meaningfully, such as connecting a person's name to their age, rather than just remembering which index holds what.

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}
print(student)
````

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}
print(len(student))
````

Here, `"name"`, `"age"`, and `"course"` are the **keys**, and `"Alice"`, `20`, and `"Data Science"` are their corresponding **values**. `len(student)` gives `3`, since it counts the number of key-value pairs, not the individual keys and values separately.

A key must always be unique within a dictionary, and is typically a string or a number. Values, on the other hand, can be of any datatype, and can even repeat.

Let's now see how to access a value using its key.

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}
print(student["name"])
````

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}
print(student["age"])
````

`student["name"]` gives us `"Alice"`, and `student["age"]` gives us `20`. Notice how we use square brackets `[ ]` with the key inside, similar to how we used index numbers with lists, except here we use a key instead of a position.

Let's see what happens when we try to access a key that doesn't exist.

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}
print(student["salary"])
````

This gives us a `KeyError: 'salary'`, since no such key exists in the dictionary. This is different from a list's `IndexError`, but the idea is similar — trying to access something that isn't there results in an error.

Now let's understand mutability. Dictionaries are mutable, similar to lists. We can update an existing value or add a completely new key-value pair after the dictionary has been created.

````python
student = {"name": "Alice", "age": 20}
student["age"] = 21
print(student)
````

````python
student = {"name": "Alice", "age": 20}
student["grade"] = "A"
print(student)
````

The first cell updates the existing key `"age"` to a new value `21`. The second cell adds a brand new key `"grade"`, which did not exist before. Both changes happened directly on the original `student` dictionary, without errors, confirming that dictionaries are mutable, just like lists.

We can iterate over a dictionary using a `for` loop too, but by default, it iterates over the keys, not the values.

````python
student = {"name": "Alice", "age": 20, "course": "Data Science"}

for key in student:
  print(key, student[key])
````

Here, `key` takes each key from the dictionary one at a time, and we use `student[key]` to fetch its corresponding value inside the loop, printing both together.

Here is a table listing the most commonly used dictionary methods, along with their syntax and arguments.

| `values()` | `dict.values()` | None | Returns all the values in the dictionary | `{"a":1,"b":2}.values()` → `dict_values([1, 2])` |
| `items()` | `dict.items()` | None | Returns all key-value pairs together | `{"a":1}.items()` → `dict_items([('a', 1)])` |

| `update()` | `dict.update(other_dict)` | `other_dict` (mandatory) | Updates the dictionary with key-value pairs from another dictionary | `{"a":1}.update({"b":2})` → `{"a":1, "b":2}` |

Let's try out each of these directly.

````python
student = {"name": "Alice", "age": 20}
print(student.keys())
````

````python
student = {"name": "Alice", "age": 20}
print(student.values())
````

````python
student = {"name": "Alice", "age": 20}
print(student.items())
````

- `keys()` returns all the keys in the dictionary.
- `values()` returns all the values in the dictionary.
- `items()` returns each key-value pair together, useful when we want both the key and value at once during iteration.

````python
student = {"name": "Alice", "age": 20}
print(student.get("salary", 0))
````

Notice how `get("salary", 0)` returns `0` instead of raising a `KeyError`, since `"salary"` does not exist in the dictionary. This is why `get()` is often safer to use than direct key access when we are unsure whether a key exists.

````python
student = {"name": "Alice", "age": 20}
student.update({"grade": "A", "age": 21})
print(student)
````

````python
student = {"name": "Alice", "age": 20}
removed_value = student.pop("age")
print(student)
print(removed_value)
````

````python
student = {"name": "Alice", "age": 20}
last_item = student.popitem()
print(student)
print(last_item)
````

````python
student = {"name": "Alice", "age": 20}
student.clear()
print(student)
````

### Comparing Strings, Lists, Tuples, and Dictionaries — In Detail

We have now studied all four data structures individually. Let's bring everything together into one detailed comparison, covering their syntax, behavior, and typical use cases.

#### Similarities

- All four are used to store collections of data rather than a single value.
- All four support the `len()` function to find out how many items they hold.
- Strings, Lists, and Tuples are all **ordered sequences**, meaning each item has a fixed position, and Dictionaries (from Python 3.7 onwards) also maintain the order in which items were inserted.
- Strings, Lists, and Tuples all support indexing, negative indexing, and slicing, since they are sequences.
- All four can be looped over using a `for` loop.
- All four can be checked for membership using the `in` and `not in` operators.

#### Detailed Differences

| Feature | String | List | Tuple | Dictionary |
|---|---|---|---|---|
| Symbol used | Quotes `" "` or `' '` | Square brackets `[ ]` | Round brackets `( )` | Curly brackets `{ }` |
| Stores | Only characters | Any datatype, mixed | Any datatype, mixed | Key-value pairs (values can be any datatype) |
| Ordered | Yes | Yes | Yes | Yes (insertion order preserved) |
| Mutable (changeable) | No | Yes | No | Yes |
| Access method | Index (position) | Index (position) | Index (position) | Key (label) |
| Supports negative indexing | Yes | Yes | Yes | No (keys are not numeric positions) |
| Supports slicing | Yes | Yes | Yes | No |
| What `len()` counts | Number of characters | Number of elements | Number of elements | Number of key-value pairs |
| Allows duplicate values | Yes | Yes | Yes | Keys must be unique; values can repeat |
| Can hold mixed datatypes | No (only characters) | Yes | Yes | Yes (for values; keys are usually string/number) |
| Can grow or shrink in size | No | Yes (append, remove, etc.) | No | Yes (add, update, remove keys) |
| Error on invalid access | `IndexError` | `IndexError` | `IndexError` | `KeyError` |
| Error on modification attempt | `TypeError` | Not applicable (allowed) | `TypeError` | Not applicable (allowed) |
| Number of built-in methods | Many (text-related) | Many (general-purpose) | Very few (`count`, `index`) | Moderate (key/value focused) |
| Common use case | Handling and processing text | General-purpose, changeable collection of items | Fixed collection that should not change | Structured or labeled data, like a record |
| Real-world example | A person's name or a sentence | A shopping list that keeps updating | Days of the week, GPS coordinates | A student's profile (name, age, marks) |

#### Quick Memory Tip

- **Strings and Tuples**: think of them as "sealed" — once written, they cannot be edited.
- **Lists and Dictionaries**: think of them as "open" — they can be edited, added to, or shrunk anytime.
- **Strings, Lists, Tuples**: think "position matters" — accessed using index numbers.

## Functions

Till now, every piece of code we wrote ran once, exactly where we wrote it. But imagine we need to do the same task at ten different places in a program, like greeting a user or calculating a total. Copying the same lines ten times is tiring, and if we later find a mistake, we have to fix it in all ten places.

This is where **functions** come in. A function is a named block of code that does one specific job. We write it once, and then we can use it as many times as we want, from anywhere.

We have actually been using functions all along: `print()`, `input()` and `len()` are all functions that Python gave us. Now let's learn to create our own.

### Defining and calling a function

We create a function using the `def` keyword. The basic syntax looks like this:

```python
def function_name():
  # block of code
```

- `def` tells Python that we are defining a function.
- `function_name` is the name we give it. The naming rules are the same as for variables.
- The `()` hold the inputs of the function (we will see them soon). They are empty for now.
- The `:` and the **indentation** mark the body of the function, just like in loops.

Defining a function does not run it. It only tells Python "remember this block of code under this name". To actually run it, we have to **call** the function by writing its name followed by `()`.

````python
def say_hello():
  print("Hello from my first function!")

say_hello()
say_hello()
````

First, we defined the function `say_hello`. Nothing was printed at that point.

Then we called it twice with `say_hello()`, and the body ran once for each call. This is the whole point of functions: write once, use many times.

Now let's see a few things that can go wrong while calling a function. First, what happens if we call a function **before** we define it? Python reads the code from top to bottom, so let's try.

````python
welcome()

def welcome():
  print("Welcome!")
````

This gives us a `NameError: name 'welcome' is not defined`. When Python reached the line `welcome()`, it had not yet executed the `def` line below it, so the name `welcome` did not exist at that moment.

The fix is simple: always define the function first, and call it afterwards.

````python
def welcome():
  print("Welcome!")

welcome()
````

Another common mistake is to forget the parentheses while calling. Let's see what that does.

````python
def say_hello():
  print("Hello from my first function!")

say_hello
print(say_hello)
````

There is no error this time, but `Hello from my first function!` was never printed either.

Without the `()`, we are only mentioning the function by its name, not calling it. Python treats the function as a value, which is why printing it shows `<function say_hello at 0x...>`. (The number after `0x` is its location in memory and will be different on every run.)

*Remember: it is the `()` that actually runs the function.*

### Parameters and arguments

A function that does the same thing every time is not very useful. Usually we want to give it some input to work with. We do this using **parameters**.

- A **parameter** is a variable name written inside the `()` while *defining* the function. It is a placeholder for a value.
- An **argument** is the actual value we pass inside the `()` while *calling* the function.

*People often use these two words interchangeably, and that is fine in daily use. Just remember that parameters are in the definition, and arguments are in the call.*

````python
def greet(name):
  print("Hello", name)

greet("Alice")
greet("Kara")
````

Here, `name` is the parameter. When we call `greet("Alice")`, the argument `"Alice"` is stored into `name` for that call, and the body prints the greeting. In the second call, `name` holds `"Kara"` instead.

A function can take more than one parameter. We separate them with commas.

````python
def add(a, b):
  print(a + b)

add(10, 20)
````

The function `add` has two parameters, `a` and `b`. In the call `add(10, 20)`, the value `10` goes into `a` and `20` goes into `b`, so `30` is printed.

Python is strict about the number of arguments. Let's see what happens if we give **fewer** values than the function needs.

````python
def greet(name):
  print("Hello", name)

greet()
````

This gives us a `TypeError` saying that `greet()` is missing 1 required positional argument: `'name'`. The function needs a value for `name`, but we did not pass any.

Now let's try giving **more** values than the function needs.

````python
def greet(name):
  print("Hello", name)

greet("Alice", "Kara")
````

This also gives us a `TypeError`, this time saying that `greet()` takes 1 positional argument but 2 were given. The function has room for only one value, so the second one has nowhere to go.

So the number of arguments in the call must match the number of parameters. (The exceptions are default values and `*args`, which we will see shortly.)

### The `return` statement

In the `add` function above, the result was only **printed** on the screen. It was shown to us, but the program itself cannot use it. For example, we cannot store the result into a variable and use it in the next calculation.

What if we want the function to hand the result back to whoever called it? That is exactly what `return` does.

**`return` sends a value back from the function to the place where it was called.** The call itself then becomes that value, so we can store it in a variable, print it, or use it in a bigger expression.

````python
def add(a, b):
  return a + b

result = add(10, 20)
print(result)
print(add(1, 2) * 10)
````

Here, `return a + b` calculates the sum and gives it back. So `add(10, 20)` becomes `30`, which is stored in `result`.

In the last line, `add(1, 2)` becomes `3` first, and then it is multiplied by `10`, giving `30`. This was not possible when the function only printed the answer.

Let's see what happens with a function which has no `return` statement.

````python
def add_and_print(a, b):
  print(a + b)

value = add_and_print(10, 20)
print(value)
````

The function printed `30` by itself. But `value` holds `None`. A function that does not return anything gives back the special value `None`, which means "nothing".

*So remember: `print` only shows a value on the screen, while `return` hands the value back to the program.*

Let's see why this matters. Suppose we try to use the answer of `add_and_print` in a further calculation.

````python
def add_and_print(a, b):
  print(a + b)

result = add_and_print(10, 20) + 5
````

The `30` got printed because the function ran. But the function gave back `None`, so Python then tried to calculate `None + 5`. This gives us a `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`, since `None` cannot be added to a number.

The fix is to use `return`, so that the call gives back a real number which can be used further.

There is one more related mistake. Let's say we calculate a value inside a function and then try to use it outside.

````python
def calculate():
  answer = 10 + 20

calculate()
print(answer)
````

This gives us a `NameError: name 'answer' is not defined`. A variable created inside a function is **local** to it. It exists only while the function is running, and disappears once the function ends. The same is true for the parameters of the function. The code outside the function cannot see them.

So how do we get a value out of a function? With `return`, of course!

````python
def calculate():
  answer = 10 + 20
  return answer

answer = calculate()
print(answer)
````

Now `calculate()` hands the value `30` back, and we store it in a variable outside, named `answer`. Note that this outside `answer` is a completely separate variable from the one inside the function. They just happen to share a name.

#### `return` ends the function

There is one more important thing about `return`. The moment Python reaches a `return` statement, the function **ends right there**. Any code written after it, inside the same function, is never executed.

````python
def test():
  print("Before return")
  return 100
  print("After return")

x = test()
print(x)
````

The first `print` runs, and then `return 100` ends the function immediately. The line `print("After return")` is skipped completely. The returned value `100` is stored in `x`.

This is similar to how `break` ends a loop. Because of this behaviour, `return` is often used to leave a function early, as in the example below.

````python
def check_age(age):
  if age < 18:
    return "Not allowed"
  return "Allowed"

print(check_age(15))
print(check_age(25))
````

For `check_age(15)`, the condition `age < 18` is `True`, so the function returns `"Not allowed"` and ends. The second `return` is never reached.

For `check_age(25)`, the condition is `False`, so Python skips the first `return` and reaches the second one, giving `"Allowed"`. We did not even need an `else` here, because the first `return` already ends the function.

A function can also return more than one value by separating them with commas. Python packs them into a tuple.

````python
def sum_and_product(a, b):
  return a + b, a * b

s, p = sum_and_product(4, 5)
print(s)
print(p)
````

The function returns two values together, and we receive them in two variables, `s` and `p`, in the same order.

### Types of arguments

Till now we passed arguments in the order the parameters were written. Python allows us to pass arguments in several ways. Let's go through them one by one.

#### Positional arguments

Positional arguments are matched to the parameters **by their position**. The first argument goes to the first parameter, the second to the second, and so on.

````python
def introduce(name, age):
  print(name, "is", age, "years old")

introduce("Alice", 20)
introduce(20, "Alice")
````

In the first call, `"Alice"` goes to `name` and `20` goes to `age`, which is what we wanted. In the second call, we changed the order, so the values went to the wrong parameters and the sentence makes no sense. Python does not complain, because it only looks at the position.

Also, the number of positional arguments must match the number of parameters. Otherwise Python raises a `TypeError`, as we saw earlier.

#### Keyword arguments

To avoid depending on the order, we can write the parameter name along with the value while calling, in the form `name=value`. These are called keyword arguments.

````python
def introduce(name, age):
  print(name, "is", age, "years old")

introduce(name="Alice", age=20)
introduce(age=20, name="Alice")
````

Both calls give the same result, even though the order is different. Python uses the names to decide which value goes where. This also makes the call easier to read.

But there are some rules. For example, a positional argument cannot come **after** a keyword argument. Let's try.

````python
def introduce(name, age):
  print(name, "is", age, "years old")

introduce(name="Alice", 20)
````

This gives us a `SyntaxError: positional argument follows keyword argument`. Python wants all positional arguments first, and keyword arguments after them. So `introduce("Alice", age=20)` is valid, but the above is not.

Another mistake is to give the same parameter a value twice.

````python
def introduce(name, age):
  print(name, "is", age, "years old")

introduce("Alice", name="Bob")
````

The first value `"Alice"` is matched by position to `name`, and then `name="Bob"` tries to set `name` again. This gives us a `TypeError: introduce() got multiple values for argument 'name'`.

The last common mistake is to misspell the parameter name while using keyword arguments.

````python
def introduce(name, age):
  print(name, "is", age, "years old")

introduce(nam="Alice", age=20)
````

This gives us a `TypeError: introduce() got an unexpected keyword argument 'nam'`. With keyword arguments, the names must match the parameter names in the function definition exactly.

#### Default arguments

We can give a parameter a **default value** while defining the function. If the caller does not pass a value for it, the default is used. If the caller does pass one, it replaces the default.

````python
def greet(name, message="Welcome"):
  print(message, name)

greet("Alice")
greet("Kara", "Good morning")
greet("Bob", message="Hello")
````

In the first call, we passed only the name, so `message` used its default value `"Welcome"`. In the second call, `"Good morning"` replaced the default. The third call shows that we can use a keyword argument to replace it as well.

There is a rule about the order though: parameters with default values must come **after** the parameters without defaults. Let's break it on purpose.

````python
def greet(message="Welcome", name):
  print(message, name)
````

This gives us a `SyntaxError`, complaining that a parameter without a default follows a parameter with a default.

Why does Python insist on this? Think about the call `greet("Alice")`. Is `"Alice"` the `message`, or the `name`? With the default parameter first, Python has no way to be sure. So it refuses the definition itself, even before any call is made. The fix is to put `name` first, which is exactly how we wrote it earlier.

#### Variable-length arguments (`*args`)

Sometimes we do not know in advance how many arguments will be passed. For example, a function that adds any amount of numbers. Writing `*args` as a parameter collects all the extra positional arguments into a **tuple**.

````python
def total(*numbers):
  print(numbers)
  return sum(numbers)

print(total(1, 2, 3))
print(total(10, 20, 30, 40, 50))
````

The name after `*` can be anything. The convention is `*args`, but a meaningful name like `*numbers` is better. Inside the function, `numbers` is a tuple holding every value passed, so we can loop over it or use functions like `sum()` on it.

#### Variable-length keyword arguments (`**kwargs`)

In the same way, writing `**kwargs` collects all the extra keyword arguments into a **dictionary**, where the argument names become the keys.

````python
def show_details(**details):
  print(details)
  for key, value in details.items():
    print(key, "->", value)

show_details(name="Alice", age=20, city="Delhi")
````

Here we passed three keyword arguments which we did not list as parameters. They all got collected into the dictionary `details`, and we looped over it to print each key and value.

We can also mix all of these in one function. The order must be: normal parameters, then default parameters, then `*args`, then `**kwargs`. In practice, you will mostly use positional, keyword and default arguments. `*args` and `**kwargs` become important when reading library code.

| Type | Example call | How values are matched |
|---|---|---|
| Positional | `introduce("Alice", 20)` | By position |
| Keyword | `introduce(age=20, name="Alice")` | By parameter name |
| Default | `greet("Alice")` | Missing value taken from the definition |
| `*args` | `total(1, 2, 3)` | Extra positional values packed into a tuple |
| `**kwargs` | `show_details(a=1, b=2)` | Extra keyword values packed into a dictionary |

### Functions for simple tasks

Now let's put functions to use. Functions are best when they are small and do one thing. Here are a few simple ones.

````python
def square(n):
  return n * n

def is_even(n):
  return n % 2 == 0

def full_name(first, last):
  return first + " " + last

print(square(7))
print(is_even(10))
print(is_even(7))
print(full_name("Alice", "Smith"))
````

- `square` returns the number multiplied by itself.
- `is_even` returns `True` or `False`. The expression `n % 2 == 0` is itself a boolean value, so we can return it directly.
- `full_name` joins two strings with a space and returns the result.

Because they return values, we can combine them freely, like `square(square(2))` or `is_even(square(3))`.

#### Maximum and minimum of three numbers

Let's write a function that finds the largest of three numbers, using the `return` statement we just learned.

````python
def maximum(a, b, c):
  if a >= b and a >= c:
    return a
  if b >= a and b >= c:
    return b
  return c

print(maximum(10, 25, 15))
print(maximum(7, 3, 2))
print(maximum(4, 4, 9))
````

The logic is as follows:

- If `a` is greater than or equal to both `b` and `c`, then `a` is the maximum, and we return it. The function ends here.
- Otherwise, we check the same for `b`. If it is the largest, we return it.
- If neither `a` nor `b` is the largest, then `c` has to be, so we simply return `c`.

Notice that we need no `else`. Since `return` ends the function, the next check is only reached if the previous one failed.

Now the minimum, with the same idea but with `<=`.

````python
def minimum(a, b, c):
  if a <= b and a <= c:
    return a
  if b <= a and b <= c:
    return b
  return c

print(minimum(10, 25, 15))
print(minimum(7, 3, 2))
print(minimum(4, 4, 9))
````

The only change is that we now check for less than or equal to. Let's use both functions together, and also compare with Python's own functions.

````python
x, y, z = 12, 45, 8

print("Max:", maximum(x, y, z))
print("Min:", minimum(x, y, z))

print("Python's max:", max(x, y, z))
print("Python's min:", min(x, y, z))
````

Python already has built-in `max()` and `min()` functions which give the same answers. We wrote our own here only to practise the logic. In real programs, prefer the built-in ones.

### Recursion

Till now, a function was called by some other code. But a function is also allowed to **call itself**. This is called **recursion**.

Recursion is useful when a problem can be broken into a smaller version of the same problem. Think of standing in a queue and wanting to know your position. You ask the person in front of you "what is your position?", who asks the person in front of them, and so on, until the first person says "I am number 1". Then each person adds one to the answer and passes it back.

Every recursive function needs two parts:

- **Base case:** the simplest situation, where we can give the answer directly without calling the function again. This is what stops the recursion.
- **Recursive case:** the part where the function calls itself with a smaller input, moving towards the base case.

> **Warning:** If there is no base case, or the input never reaches it, the function keeps calling itself until Python stops it with a `RecursionError`. It is the same danger as an infinite loop.

Let's understand recursion by solving two classic problems, first with loops, and then with recursion.

#### Factorial

The factorial of a number `n`, written `n!`, is the product of all whole numbers from `1` to `n`. For example, `5! = 5 × 4 × 3 × 2 × 1 = 120`. By definition, `0! = 1`.

**Using a loop**

We start with a result of `1` and multiply it by each number from `1` to `n`.

````python
def factorial_loop(n):
  result = 1
  for i in range(1, n + 1):
    result = result * i
  return result

print(factorial_loop(5))
print(factorial_loop(0))
````

`result` starts at `1`. The loop runs for `i = 1, 2, 3, 4, 5`, and each time `result` is multiplied by `i`. So it becomes `1, 2, 6, 24, 120`. For `n = 0`, `range(1, 1)` is empty, so the loop does not run and `1` is returned, which matches `0! = 1`.

**Using recursion**

Notice that `5! = 5 × 4!`. In general, `n! = n × (n-1)!`. So the factorial of `n` is defined using the factorial of a smaller number. That is a perfect fit for recursion.

- Base case: if `n` is `0`, the answer is `1`.
- Recursive case: the answer is `n * factorial_recursive(n - 1)`.

````python
def factorial_recursive(n):
  if n == 0:
    return 1
  return n * factorial_recursive(n - 1)

print(factorial_recursive(5))
print(factorial_recursive(0))
````

Let's trace `factorial_recursive(3)` to see what happens:

```text
factorial_recursive(3)
= 3 * factorial_recursive(2)
= 3 * (2 * factorial_recursive(1))
= 3 * (2 * (1 * factorial_recursive(0)))
= 3 * (2 * (1 * 1))      <- base case reached, returns 1
= 3 * (2 * 1)
= 3 * 2
= 6
```

The function keeps calling itself with a smaller number until `n` becomes `0`. That is the base case, which returns `1`. Then each waiting call receives the answer it was waiting for, multiplies it, and returns it to the call before it. This is where `return` does the real work.

Now let's see what happens if we forget the base case. Here is the same function, without the `if n == 0` check.

````python
def factorial_broken(n):
  return n * factorial_broken(n - 1)

factorial_broken(5)
````

This gives us a `RecursionError: maximum recursion depth exceeded`.

Let's see why. `factorial_broken(5)` calls `factorial_broken(4)`, which calls `factorial_broken(3)`, and so on. After `0`, it does not stop but goes on to `-1`, `-2`, `-3`... There is no base case to ever tell it "stop, you have the answer". Every call that is waiting also takes up some memory, so Python protects itself by allowing only a limited number of nested calls (about 1000 by default). When that limit is crossed, it stops the program with a `RecursionError`.

This is why the base case is not optional. The correct version is the `factorial_recursive` function we wrote above.

#### Fibonacci

The Fibonacci series is a sequence where each number is the sum of the previous two numbers, starting with `0` and `1`:

```text
0, 1, 1, 2, 3, 5, 8, 13, 21, ...
```

We number the terms from `0`. So `fib(0) = 0`, `fib(1) = 1`, `fib(2) = 1`, `fib(3) = 2`, `fib(4) = 3`, and so on.

**Using a loop**

We keep track of the last two numbers and keep moving them forward.

````python
def fibonacci_loop(n):
  a, b = 0, 1
  for i in range(n):
    a, b = b, a + b
  return a

print(fibonacci_loop(0))
print(fibonacci_loop(1))
print(fibonacci_loop(7))

for i in range(10):
  print(fibonacci_loop(i), end=" ")
````

`a` and `b` start as the first two terms, `0` and `1`. In each round of the loop, the line `a, b = b, a + b` shifts them forward: the new `a` is the old `b`, and the new `b` is the sum of the old two. After `n` rounds, `a` holds the `n`th term.

The `end=" "` inside `print` makes the output stay on the same line with a space, instead of moving to a new line.

**Using recursion**

The definition itself is recursive: each term is the sum of the previous two, `fib(n) = fib(n-1) + fib(n-2)`.

- Base cases: `fib(0) = 0` and `fib(1) = 1`.
- Recursive case: `fib(n) = fib(n - 1) + fib(n - 2)`.

````python
def fibonacci_recursive(n):
  if n == 0:
    return 0
  if n == 1:
    return 1
  return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

print(fibonacci_recursive(0))
print(fibonacci_recursive(1))
print(fibonacci_recursive(7))

for i in range(10):
  print(fibonacci_recursive(i), end=" ")
````

We get the same results as the loop version. Here there are two base cases, because each term depends on two earlier terms. Also, this function calls itself **twice** in one go, so the calls branch out like a tree.

*Because of that branching, the recursive Fibonacci repeats the same calculations many times (for example, `fibonacci_recursive(5)` calculates `fibonacci_recursive(3)` more than once). For small `n` this is fine, but for something like `n = 40` it becomes very slow, while the loop version is still instant.*

#### Loop or recursion?

| | Loop | Recursion |
|---|---|---|
| Idea | Repeat steps and update variables | A function solves a smaller copy of itself |
| Needs | A loop condition or sequence | A base case and a recursive case |
| Risk | Infinite loop | `RecursionError` if the base case is never reached |
| Speed and memory | Usually faster and lighter | Each call takes extra memory, can be slower |
| Best for | Simple repetition | Problems that naturally split into smaller versions of themselves |

Anything that can be done with recursion can also be done with a loop. We learn recursion because some problems, like working with trees and nested data, become much simpler to express with it.
