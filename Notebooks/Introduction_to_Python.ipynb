Python is preferred in Data Science and AI for several structural reasons:

- Simple, readable syntax that resembles plain English, reducing the learning curve for non-programmers like statisticians and domain experts

- Strong community support, meaning most problems already have documented solutions or forum discussions
- Interpreted nature, allowing quick testing and iteration without a compile step
- Cross-platform compatibility, so the same code runs on Windows, macOS, and Linux
- Seamless integration with other languages (C, C++, Java) for performance-critical tasks

### Python vs Traditional Programming Languages

| Aspect | Python | Traditional Languages (C/Java) |
|---|---|---|
| Syntax complexity | Minimal, close to pseudocode | Verbose, strict typing rules |
| Development speed | Fast prototyping | Slower due to boilerplate code |
| Library support for ML/AI | Extensive and mature | Limited or requires more setup |
| Community and resources | Massive, active | Smaller in the AI/ML domain |
| Execution speed | Slower (interpreted) | Faster (compiled) |

### Datatypes - Part 1

Before going any further, we will discuss few important things. One very important thing is datatype.

Every value that we use in python or any other programming language has a datatype.

Datatype means the type of data - whether it is an number, a string, or a boolean value etc. Based on the type of data, we have different valid operations or functions that we can perform on that data.

Let's try to understand that you have 10 apples. Here the number 10 is an integer. Now, your friend has 20 apples. Here 20 is an integer too. Certainly you are able to add 10 and 20, and get 30 apples. You were able to add those numbers.

Now, if your name is Alice and your friend's name is Kara, are you able to sum those two names? No!

Another example is that, a question is asked to you and your friend that whether you have completed the assignments or not. You say `Yes` and your friend says `No`. What can be the valid operation on such values of yes and no?

Let us see the basic ones for now. Try to follow the names as given here. There are named as so in python and we have to follow that.

For numeric datatypes, we generally have:

- Integer: Denoted as `int`. Integer values. Example: 10, -20, 30, 0 etc

- Float: Denoted as `float`. Numeric values with the dot! Example: 10.1, -23.0, 3.14, 2.22 etc

- Complex Numbers also exist but very little application in this course.

For alphanumerics and other symbols we have:

- Character: One single symbol by itself. Example: A, B, c, d, 1, #, @ etc. (All of these needs to be enclosed in quotes.)

- String: Denoted by `str`. Sequence of character(s). Example: ABcd, A#2124, Alice, 132324 etc. (All of these needs to be enclosed in quotes.)



### Connecting data, datatypes and variables

We were using various data till now. We took input and used print. We stored values in variables too. It is important to understand that every value has its datatype and therefore its valid operations too.

Let's understand with examples now.

````python
var1 = 10
````

The 10 here is of the type integer and valid operations are as per its datatype (we will discuss that later).

````python
var1 = 10.0
````

The 10 here is written as 10.0 which means we are using the float datatype and its a numeric 10.0

````python
var1 = "10"
var2 = "10.0"
````

Here both the variables have been assigned two values which are strings since they are enclosed in quotes. We will not worry about any numeric similarity here. For programming, the strings `"10"` and `"10.0"` are just separate values.

The important fact is that in Python, a programmer (you) do not need to **explicitly** mention the datatype of a variable which needs to be done in other programming languages.

Infact, in Python, we have **dynamic assignment**, which means that we assign an integer to a variable at some point, and later assign some value of different datatype without any issue. Let's see an example.

````python
var1 = 100
print("Value is: ", var1)
print("Type is: ", type(var1))

# reassign different datatype value now

var1 = "Alice"
print("Value is: ", var1)
print("Type is: ", type(var1))
````

From the output it should be clear that `var1` had stored an interger 100 before. The `type()` function confirms the datatype too.

For the 2nd time, we have assigned a string `"Alice"` to `var1`. The datatype the variable is automatically adjusted by python internally. Also we can confirm the string datatype using the `type()` function.

Also at this point it should be clear, that changing the value of a variable is definitely possible since the name itself is variable. So the below example should be naive to include here:

````python
var1 = 10
print(var1)

# change value

var1 = 20
print(var1)
````

Now, let's understand that user input can be stored into variables. Let's try that with some examples and explanations.

````python
name = input("Enter your name: ")
age = input("Enter your age: ")

# let us print the name and age to see if we have the values saved

print("Name:", name)
print("Age:", age)
````

Well that works! But let's run the program again and see the datatypes of them once. As said earlier, input function takes inputs in string format by default. So both of them should be strings.

````python
name = input("Enter your name: ")
age = input("Enter your age: ")

# let us print the name and age to see if we have the values saved

print("Name:", name)
print("Type of name:", type(name))
print("Age:", age)
print("Type of age:", type(age))
````

Well, both are in string. But certainly you will want the name to be a string, and the age to be of an integer type (don't bother about negative or zero age). Let's try to do that and understand via code.

````python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

# let us print the name and age to see if we have the values saved

print("Name:", name)
print("Type of name:", type(name))
print("Age:", age)
print("Type of age:", type(age))
````

We used something called **explicit type casting** here. It means changing the datatype of something into a datatype that we want. Since a string `20` has a valid integer representation, we can convert it without issues. But remember the same cannot be done with some value like `Alice`.

The `int(...)` typecasted the string 20 into integer value 20 which is confirmed from the type output.
