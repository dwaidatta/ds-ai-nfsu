## Q3

```python
score = 78
attendance = 85

if score >= 90:
    grade = "A"
elif score >= 75:
    if attendance >= 80:
        grade = "B+"
    else:
        grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(grade)
```

```python
a = 10
b = 3
c = 7

condition1 = (a - b) > c
condition2 = (a % b == 1) or (c < b)
condition3 = not (a < c)

if condition1 and condition2:
    print("Path A")
elif condition3 or condition2:
    print("Path B")
else:
    print("Path C")
```
