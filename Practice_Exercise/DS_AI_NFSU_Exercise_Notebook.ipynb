````python
# @title Q4
p = 12
q = 4
r = 6

flag1 = (p / q == 3) and (r % q != 0)
flag2 = not (p < r)

if flag1 and flag2:
    print("Alpha")
elif flag1 or flag2:
    print("Beta")
else:
    print("Gamma")
````

````python
# @title Q5
total = 0
for i in range(1, 10):
    if i % 3 == 0:
        total += i
    elif i % 2 == 0:
        total -= i

print(total)
````

````python
# @title Q6
result = 0
for i in range(1, 4):
    j = 0
    while j < 3:
        j += 1
        if i == j and i % 2 != 0:
            result += i * j
print(result)
````

Note: The question below uses the `+` operators in Strings. When `+` is used in Strings, we call the operation as **concatenation**. The characters/strings just get attached with the existing string/character.

Here is an example to understand.

```python
s = "A"
s = s + "B"
```

That makes
```python
s = "AB"
```

Now, another example,

```python
# empty string
s = ""

s = s + "ABCD"
```

That makes,

```python
s = "ABCD"
```
